```python
import heapq

# Goal state
GOAL = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

# Initial state
INITIAL = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)


def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()



def manhattan_distance(state):

    distance = 0

    for i in range(9):

        tile = state[i]

        # Don't calculate distance for blank
        if tile == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Goal position
        goal_index = GOAL.index(tile)

        goal_row = goal_index // 3
        goal_col = goal_index % 3

        # Manhattan distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance



def get_neighbors(state):

    neighbors = []

    # Find blank
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


def a_star():


    open_list = []

    h = manhattan_distance(INITIAL)

    heapq.heappush(
        open_list,
        (h, 0, INITIAL)
    )


    g_cost = {
        INITIAL: 0
    }


    parent = {
        INITIAL: None
    }

    while open_list:

        f, g, current = heapq.heappop(open_list)


        if g != g_cost[current]:
            continue


        if current == GOAL:


            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\n===== SOLUTION =====")
            print("Total moves:", len(path) - 1)

            for step, state in enumerate(path):

                g_value = step
                h_value = manhattan_distance(state)
                f_value = g_value + h_value

                print("\nStep:", step)
                print("g(n) =", g_value)
                print("h(n) =", h_value)
                print("f(n) =", f_value)

                print_puzzle(state)

            return

        for neighbor in get_neighbors(current):

            new_g = g + 1


            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g
                parent[neighbor] = current

                h = manhattan_distance(neighbor)
                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, neighbor)
                )

    print("No solution found.")



print("===== A* SEARCH =====")
print("Heuristic: Manhattan Distance")

print("\nInitial State:")
print_puzzle(INITIAL)

print("Goal State:")
print_puzzle(GOAL)

a_star()
```
