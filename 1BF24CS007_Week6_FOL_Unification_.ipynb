```python
import re

def is_variable(x):
    """Determines if a term is a variable (lowercase string)."""
    return isinstance(x, str) and x.islower() and len(x) == 1

def occurs_check(var, term):
    """Prevents infinite loops by checking if a variable is inside a term."""
    if var == term:
        return True
    if isinstance(term, list):
        return any(occurs_check(var, t) for t in term)
    return False

def apply_subst(subst, expr):
    """Applies the current substitutions to an expression."""
    if not subst:
        return expr
    if isinstance(expr, str):
        return subst.get(expr, expr)
    if isinstance(expr, list):
        return [apply_subst(subst, arg) for arg in expr]
    return expr

def parse_expression(text):
    """
    Parses a string like 'Eats(x, Apple)' into a list format ['Eats', 'x', 'Apple'].
    Also handles flat lists or tokens separated by spaces/commas.
    """
    text = text.strip()
    # Match standard predicate format: Predicate(arg1, arg2, ...)
    match = re.match(r"^([A-Za-z0-9_]+)\((.*)\)$", text)
    if match:
        predicate = match.group(1)
        args_str = match.group(2)
        # Split arguments by comma and strip whitespace/quotes
        args = [arg.strip().strip('"').strip("'") for arg in args_str.split(',') if arg.strip()]
        return [predicate] + args
    else:
        # Fallback if user types space-separated words
        return [token.strip('"').strip("'") for token in text.replace(',', ' ').split() if token]

def format_expression(expr):
    """Converts the internal list representation back to a readable string."""
    if isinstance(expr, list) and len(expr) > 0:
        if len(expr) == 1:
            return f"{expr[0]}()"
        return f"{expr[0]}({', '.join(str(e) for e in expr[1:])})"
    return str(expr)

def unify(psi1, psi2, subst=None):
    """Main Unification Algorithm execution."""
    if subst is None:
        subst = {}

    if isinstance(psi1, str) and isinstance(psi2, str):
        if psi1 == psi2:
            return subst
        if is_variable(psi1):
            if occurs_check(psi1, psi2): return "FAILURE"
            subst[psi1] = psi2
            return subst
        if is_variable(psi2):
            if occurs_check(psi2, psi1): return "FAILURE"
            subst[psi2] = psi1
            return subst
        return "FAILURE"

    if is_variable(psi1):
        if occurs_check(psi1, psi2): return "FAILURE"
        subst[psi1] = psi2
        return subst
    if is_variable(psi2):
        if occurs_check(psi2, psi1): return "FAILURE"
        subst[psi2] = psi1
        return subst

    if isinstance(psi1, list) and isinstance(psi2, list):
        if psi1[0] != psi2[0]:
            return "FAILURE"
        if len(psi1) != len(psi2):
            return "FAILURE"

        for i in range(1, len(psi1)):
            arg1 = apply_subst(subst, psi1[i])
            arg2 = apply_subst(subst, psi2[i])

            S = unify(arg1, arg2, subst)
            if S == "FAILURE":
                return "FAILURE"
            subst.update(S)

        return subst

    return "FAILURE"

# --- User Interaction Loop ---
print("=== First-Order Logic Unifier ===")
print("Rules:")
print("1. Variables must be a single lowercase letter (e.g., x, y, z)")
print("2. Constants or Predicates should start with a Capital letter (e.g., Riya, Apple, Eats)")
print("3. Format example: Eats(x, Apple)\n")

while True:
    input_A = input("Enter Expression A (or 'exit' to quit): ")
    if input_A.lower() == 'exit':
        break

    input_B = input("Enter Expression B: ")

    expr_A = parse_expression(input_A)
    expr_B = parse_expression(input_B)

    print(r"Parsed Internal Form A:", expr_A)
    print(r"Parsed Internal Form B:", expr_B)

    # Run the unification algorithm
    # We copy the dictionary to keep executions fresh
    result = unify(expr_A, expr_B, subst={})

    print("\n--- Result ---")
    if result == "FAILURE":
        print("❌ Unification Failed! The expressions cannot match.")
    else:
        print("✅ Success! Substitutions applied:")
        if not result:
            print("  None (Expressions are already identical)")
        for var, val in result.items():
            print(f"  {var} -> {val}")

        # Display unified string
        unified_output = apply_subst(result, expr_A)
        print(f"Unified Form: {format_expression(unified_output)}")
    print("-" * 40 + "\n")
```

```python
def is_variable(x):
    return isinstance(x, str) and x.islower() and len(x) == 1

def occurs_check(var, term):
    if var == term:
        return True
    if isinstance(term, list):
        return any(occurs_check(var, t) for t in term)
    return False

def apply_subst(subst, expr):
    if not subst:
        return expr
    if isinstance(expr, str):
        return subst.get(expr, expr)
    if isinstance(expr, list):
        return [apply_subst(subst, arg) for arg in expr]
    return expr

def parse_expression(text):
    text = text.replace(" ", "")
    if '(' not in text:
        return text

    idx = text.find('(')
    name = text[:idx]
    args_str = text[idx+1:-1] if text.endswith(')') else text[idx+1:]

    args = []
    current = []
    paren_depth = 0

    for char in args_str:
        if char == ',' and paren_depth == 0:
            args.append(parse_expression("".join(current)))
            current = []
        else:
            if char == '(':
                paren_depth += 1
            elif char == ')':
                paren_depth -= 1
            current.append(char)

    if current:
        args.append(parse_expression("".join(current)))

    return [name] + args

def format_expression(expr):
    if isinstance(expr, list):
        if len(expr) == 1:
            return f"{expr}()"
        return f"{expr}({', '.join(format_expression(e) for e in expr[1:])})"
    return str(expr)

def unify(psi1, psi2, subst=None):
    if subst is None:
        subst = {}

    if isinstance(psi1, str) and isinstance(psi2, str):
        if psi1 == psi2:
            return subst
        if is_variable(psi1):
            if occurs_check(psi1, psi2): return "FAILURE"
            subst[psi1] = psi2
            return subst
        if is_variable(psi2):
            if occurs_check(psi2, psi1): return "FAILURE"
            subst[psi2] = psi1
            return subst
        return "FAILURE"

    if is_variable(psi1):
        if occurs_check(psi1, psi2): return "FAILURE"
        subst[psi1] = psi2
        return subst
    if is_variable(psi2):
        if occurs_check(psi2, psi1): return "FAILURE"
        subst[psi2] = psi1
        return subst

    if isinstance(psi1, list) and isinstance(psi2, list):
        if psi1[0] != psi2[0]:
            return "FAILURE"
        if len(psi1) != len(psi2):
            return "FAILURE"

        for i in range(1, len(psi1)):
            arg1 = apply_subst(subst, psi1[i])
            arg2 = apply_subst(subst, psi2[i])

            S = unify(arg1, arg2, subst)
            if S == "FAILURE":
                return "FAILURE"
            subst.update(S)

        return subst

    return "FAILURE"



while True:
    input_A = input("Enter Expression A (or 'exit' to quit): ")
    if input_A.lower() == 'exit':
        break

    input_B = input("Enter Expression B: ")

    expr_A = parse_expression(input_A)
    expr_B = parse_expression(input_B)

    print("\n[Parsed Structure A]:", expr_A)
    print("[Parsed Structure B]:", expr_B)

    result = unify(expr_A, expr_B, subst={})

    print("\n--- Result ---")
    if result == "FAILURE":
        print("Unification Failed!")
    else:
        print("Success! Substitutions applied:")
        if not result:
            print("  None")
        for var, val in result.items():
            print(f"  {var} -> {format_expression(val)}")

        unified_output = apply_subst(result, expr_A)
        print(f"\nUnified Form: {format_expression(unified_output)}")
    print("-" * 50 + "\n")
```
