```python
#tic tac toe game

board = [' ' for i in range(9)]

def display():
    print()
    print(board[0], '|', board[1], '|', board[2])
    print('--+---+--')
    print(board[3], '|', board[4], '|', board[5])
    print('--+---+--')
    print(board[6], '|', board[7], '|', board[8])
    print()


def check_winner(player):


    if board[0] == board[1] == board[2] == player:
        return True
    if board[3] == board[4] == board[5] == player:
        return True
    if board[6] == board[7] == board[8] == player:
        return True


    if board[0] == board[3] == board[6] == player:
        return True
    if board[1] == board[4] == board[7] == player:
        return True
    if board[2] == board[5] == board[8] == player:
        return True


    if board[0] == board[4] == board[8] == player:
        return True
    if board[2] == board[4] == board[6] == player:
        return True

    return False


player = 'X'

for turn in range(9):

    display()

    position = int(input("Player " + player + ", enter position (1-9): "))

    position = position - 1

    if board[position] != ' ':
        print("Position already occupied!")
        continue

    board[position] = player

    if check_winner(player):
        display()
        print("Player", player, "wins!")
        break

    if player == 'X':
        player = 'O'
    else:
        player = 'X'

else:
    display()
    print("Game Draw!")
```

```python
class VacuumCleaner:
    def __init__(self, location, rooms):
        self.location = location
        self.rooms = rooms

    def perceive_and_act(self):

        if self.rooms[self.location] == "Dirty":
            print(f"Vacuum is in Room {self.location} - Room is Dirty")
            print("Action: SUCK")
            self.rooms[self.location] = "Clean"

        else:
            print(f"Vacuum is in Room {self.location} - Room is Clean")

            if self.location == "A":
                print("Action: MOVE RIGHT (A -> B)")
                self.location = "B"
            else:
                print("Action: MOVE LEFT (B -> A)")
                self.location = "A"

        print("Current state:", self.rooms)
        print("-" * 40)

rooms = {
    "A": "Dirty",
    "B": "Dirty"
}

vacuum = VacuumCleaner("A", rooms)

for i in range(6):
    vacuum.perceive_and_act()
```

```python
class GoalBasedVacuumCleaner:
    def __init__(self, location, rooms):
        self.location = location
        self.rooms = rooms

    def goal_achieved(self):

        return all(status == "Clean" for status in self.rooms.values())

    def choose_action(self):

        if self.rooms[self.location] == "Dirty":
            return "SUCK"

        if self.location == "A":
            other_room = "B"
        else:
            other_room = "A"

        if self.rooms[other_room] == "Dirty":
            if self.location == "A":
                return "MOVE RIGHT"
            else:
                return "MOVE LEFT"
        return "STOP"

    def execute_action(self, action):

        if action == "SUCK":
            print(f"Room {self.location} is Dirty")
            print("Action: SUCK")
            self.rooms[self.location] = "Clean"

        elif action == "MOVE RIGHT":
            print("Action: MOVE RIGHT (A -> B)")
            self.location = "B"

        elif action == "MOVE LEFT":
            print("Action: MOVE LEFT (B -> A)")
            self.location = "A"

        elif action == "STOP":
            print("Goal achieved! All rooms are clean.")

        print("Current location:", self.location)
        print("Environment:", self.rooms)
        print("-" * 40)



rooms = {
    "A": "Dirty",
    "B": "Dirty"
}


vacuum = GoalBasedVacuumCleaner("A", rooms)

while not vacuum.goal_achieved():

    action = vacuum.choose_action()
    vacuum.execute_action(action)

print("Final Environment:", vacuum.rooms)
print("Final Location:", vacuum.location)
```

```python
n = int(input("Enter number of rooms: "))

rooms = []

# Input room status
for i in range(n):
    status = int(input(f"Enter status of Room {i + 1} (1=Dirty, 0=Clean): "))
    rooms.append(status)

print("\nInitial Room Status:", rooms)

# Goal: All rooms must be clean
goal = [0] * n

print("\n--- Vacuum Cleaner Agent ---")

# Visit each room
for i in range(n):

    print(f"\nVacuum is in Room {i + 1}")

    # Simple Reflex Agent
    if rooms[i] == 1:
        print("Room is Dirty -> Clean the room")
        rooms[i] = 0
    else:
        print("Room is Clean -> Move to next room")

    # Check Goal
    if rooms == goal:
        print("\nGoal Achieved: All rooms are clean!")
        break

# Final state
print("\nFinal Room Status:", rooms)

if rooms == goal:
    print(" Goal achieved successfully.")
else:
    print("Goal-Based Agent: Goal not achieved.")
```
